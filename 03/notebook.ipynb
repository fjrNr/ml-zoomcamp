```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mutual_info_score
```

Data Preparation
---
---

```python
df = pd.read_csv('course_lead_scoring_2026.csv')
df.head()
```

```python
# Check missing values
df.isna().sum()
```

```python
# Fill missing values in categorial features
categorical_features = df.select_dtypes(include=['object']).columns
for column in categorical_features:
    df[column] = df[column].fillna('NA')

df.isna().sum()
```

```python
# Fill missing values in numeric features
numeric_features = df.select_dtypes(include=['float']).columns
for column in numeric_features:
    df[column] = df[column].fillna(0.0)

df.isna().sum()
```

Q3. The biggest mutual information score with `converted`
---
---

```python
def mutual_info_converted_score(series):
    return mutual_info_score(series, df_full_train.converted)

mutual_info_all = df_full_train[categorical_features].apply(mutual_info_converted_score)
mutual_info_all.sort_values(ascending=False)
```

The biggest mutual information score with `converted` is `lead_source`
